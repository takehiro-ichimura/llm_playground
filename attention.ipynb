```python
import torch

inputs = torch.tensor([
    [0.43, 0.15, 0.89], # Your (x^1)
    [0.55, 0.87, 0.66], # journey (x^2)
    [0.57, 0.85, 0.64], # starts (x^3)
    [0.22, 0.58, 0.33], # with (x^4)
    [0.77, 0.25, 0.10], # one (x^5)
    [0.05, 0.80, 0.55] # step (x^6)
])
```

```python
query = inputs[1] # jpurney (x^2)

atten_scores_2 = torch.empty(inputs.shape[0])  # 各単語ベクトルとクエリベクトルの内積を計算
for i, x_i in enumerate(inputs):
    atten_scores_2[i] = torch.dot(x_i, query)
print(atten_scores_2)
```

```python
attn_weights_2_tmp = atten_scores_2 / atten_scores_2.sum()
print(attn_weights_2_tmp)
print("Sum: ", attn_weights_2_tmp.sum())
```

```python
def softmax_naive(x):
    return torch.exp(x) / torch.exp(x).sum(dim=0)

attn_weights_2_naive = softmax_naive(atten_scores_2)
print(attn_weights_2_naive)
print("Sum: ", attn_weights_2_naive.sum())
```

```python
attn_weights_2 = torch.softmax(atten_scores_2, dim=0)
print(attn_weights_2)
print("Sum: ", attn_weights_2.sum())
```

```python
attn_scores = torch.empty(6, 6)
for i, x_i in enumerate(inputs):
    for j, x_j in enumerate(inputs):
        attn_scores[i, j] = torch.dot(x_i, x_j)
print(attn_scores)
```

```python
attn_scores = inputs @ inputs.T
print(attn_scores)
```

```python
attn_weights = torch.softmax(attn_scores, dim=1)
print(attn_weights)
```

```python
row_2_sum = sum([0.1385, 0.2379, 0.2333, 0.1240, 0.1082, 0.1581])
print("Row 2 sum:", row_2_sum)
print("all row sum:", attn_weights.sum(dim=1))
```

```python
all_context_vectors = attn_weights @ inputs
print(all_context_vectors)
```

```python
# Self Attention

x_2 = inputs[1]  # journey (x^2)
d_in = inputs.shape[1]
d_out = 2
```

```python
torch.manual_seed(123)
W_query = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
W_key = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
W_value = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)

query_2 = x_2 @ W_query  # クエリベクトル
key_2 = x_2 @ W_key     # キーベクトル
value_2 = x_2 @ W_value # バリューベクトル

print("Query vector:", query_2)
print("Key vectors:", key_2)
print("Value vectors:", value_2)
```

```python
keys = inputs @ W_key
values = inputs @ W_value
print("All key vectors:", keys)
print("All value vectors:", values)
```

```python
keys_2 = keys[1]  # journey (x^2)
attn_score_22 = query_2.dot(keys_2)
print("Attention score (x^2, x^2):", attn_score_22)
```

```python
attn_scores_2 = query_2 @ keys.T
print("Attention scores for x^2:", attn_scores_2)
```

```python
d_k = keys.shape[-1]
atttn_weights_2 = torch.softmax(attn_scores_2 / d_k**0.5, dim=-1)
print("Attention weights for x^2:", atttn_weights_2)
```

```python
context_vec_2 = attn_weights_2 @ values
print("Context vector for x^2:", context_vec_2)
```
